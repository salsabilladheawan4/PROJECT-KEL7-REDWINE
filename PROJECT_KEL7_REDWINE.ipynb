```python
# Mengimpor library yang dibutuhkan
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Mengatur gaya grafik agar terlihat lebih rapi dan profesional
sns.set_theme(style="whitegrid")
print("Library berhasil di-import!")
```

```python
# Membaca dataset excel
file_path = "3TIE_07_RedWineQuality.xlsx"
df = pd.read_excel(file_path)
```

```python
print("5 Baris Pertama Dataset:")
display(df.head())
```

```python
print("--- Ringkasan Statistika Deskriptif Keseluruhan ---")
display(df.describe())
```

```python
print("\n")
```

```python
print("--- Statistika Deskriptif Spesifik untuk Kandungan Alkohol (alcohol) ---")
print(f"Rata-rata (Mean)      : {df['alcohol'].mean():.2f}")
print(f"Nilai Tengah (Median) : {df['alcohol'].median():.2f}")
print(f"Nilai Sering (Modus)  : {df['alcohol'].mode()[0]:.2f}")
print(f"Nilai Terendah (Min)  : {df['alcohol'].min():.2f}")
print(f"Nilai Tertinggi (Max) : {df['alcohol'].max():.2f}")
print(f"Standar Deviasi (Std) : {df['alcohol'].std():.2f}")
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='quality', palette='viridis')
```

```python
# Membuat 2 grafik berdampingan untuk Alcohol dan pH
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histogram untuk Alcohol
sns.histplot(df['alcohol'], bins=20, kde=True, color='blue', ax=axes[0])
axes[0].set_title('Distribusi Kandungan Alkohol', fontweight='bold')
axes[0].set_xlabel('Alkohol (%)')
axes[0].set_ylabel('Frekuensi')

# Histogram untuk pH
sns.histplot(df['pH'], bins=20, kde=True, color='green', ax=axes[1])
axes[1].set_title('Distribusi Tingkat Keasaman (pH)', fontweight='bold')
axes[1].set_xlabel('pH')
axes[1].set_ylabel('Frekuensi')

plt.tight_layout()
plt.show()
```

```python
plt.title('Distribusi Kualitas Red Wine', fontsize=14, fontweight='bold')
plt.xlabel('Tingkat Kualitas (Quality)', fontsize=12)
plt.ylabel('Jumlah Sampel', fontsize=12)

# Menampilkan grafik
plt.show()
```

```python
# Membuat 2 grafik berdampingan untuk Alcohol dan pH
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histogram untuk Alcohol
sns.histplot(df['alcohol'], bins=20, kde=True, color='blue', ax=axes[0])
axes[0].set_title('Distribusi Kandungan Alkohol', fontweight='bold')
axes[0].set_xlabel('Alkohol (%)')
axes[0].set_ylabel('Frekuensi')

# Histogram untuk pH
sns.histplot(df['pH'], bins=20, kde=True, color='green', ax=axes[1])
axes[1].set_title('Distribusi Tingkat Keasaman (pH)', fontweight='bold')
axes[1].set_xlabel('pH')
axes[1].set_ylabel('Frekuensi')

plt.tight_layout()
plt.show()
```

```python
# Membuat 2 Boxplot berdampingan
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Boxplot untuk Fixed Acidity
sns.boxplot(data=df, y='fixed acidity', color='orange', ax=axes[0])
axes[0].set_title('Boxplot Fixed Acidity (Cek Outlier)', fontweight='bold')

# Boxplot untuk Residual Sugar
sns.boxplot(data=df, y='residual sugar', color='red', ax=axes[1])
axes[1].set_title('Boxplot Residual Sugar (Cek Outlier)', fontweight='bold')

plt.tight_layout()
plt.show()
```
